### Funções
Uma função é uma **receita com nome**: você entrega valores (os **ingredientes**), ela faz as contas e **devolve** um resultado (o **bolo**). Quem usa a função só precisa saber o **nome** e **o que entregar**: não precisa lembrar o passo a passo.

Para **criar** uma função usamos `def`, e para **devolver** o resultado usamos `return`. Para **usar**, escrevemos o nome e, entre parênteses, o que entregamos. Isso se chama **chamar** a função.

````python
def dobro(numero):
    return numero * 2

print(dobro(4))
print(dobro(10))
````

## Bloco 2 — Transform: limpar a base suja

Esta é a parte mais importante do trabalho com dados. Uma frase para guardar o dia inteiro:

> **uma base de dados boa vale mais do que um modelo sofisticado.**

Se a base tiver música repetida, texto escrito de qualquer jeito ou valores impossíveis, **nenhuma IA vai consertar isso depois**.

Já olhamos a base e achamos quatro tipos de sujeira. Vamos limpar um de cada vez, nesta ordem:
1. **Campos vazios**: tirar a linha ou preencher
2. **Textos bagunçados**: maiúsculas, minúsculas e espaços sobrando
3. **Linhas repetidas**: a mesma música duas vezes
4. **Valores impossíveis**: bpm e duração zerados

### Passo 1 — Campos vazios

Um campo vazio aparece como `NaN` na tabela. Existem **duas** saídas, e escolher entre elas é uma decisão sua:

- **Apagar a linha**, quando a informação é essencial. Uma música sem título ou sem artista não serve para o app: `dados.dropna(subset=['coluna1', 'coluna2'])`.
- **Preencher com um valor padrão**, quando a informação é um enfeite. Um álbum vazio não impede a música de tocar: `dados['coluna'].fillna('texto padrão')`.

> Atenção: nem todo vazio é erro. A coluna `regiao` fica vazia para artistas de fora do Brasil, e isso está certo. Vamos deixá-la como está.

| `.str.upper()` | põe tudo em maiúscula | `'br'` vira `'BR'` |

### O histograma: quantas músicas em cada faixa?

Imagine que você separa todas as músicas em **caixinhas**, de acordo com a energia: a caixinha de energia 0,0 a 0,1, a de 0,1 a 0,2, e assim por diante. Depois você **conta quantas músicas caíram em cada caixinha** e desenha uma barra com essa altura. Isso é um **histograma**.

- **Barra alta** = muita música naquela faixa.
- **Barra baixa** = pouca música naquela faixa.

````python
plt.figure(figsize=(6, 4))
plt.hist(musicas['energia'], bins=30)
plt.title('energia')
plt.xlabel('valor (0 a 1)')
plt.ylabel('nº de músicas')
plt.show()
````

### Parte 2 — E se a pessoa pedir outra coisa?

A sua playlist de treino ficou ótima. Mas ela só serve para **treino**. Se a pessoa pedir **estudar**, você teria que refazer tudo, trocando os números e as palavras à mão.

Agora vem a ideia central do dia: **deixar o computador fazer isso por nós**. A pessoa escreve um **pedido** (o dicionário que você criou no Bloco 0, com `energia`, `ocasiao`, `humor`...) e uma **função** pronta, chamada `recomendar`, monta a playlist.

Há uma diferença em relação ao que você fez à mão. A peneira à mão é **rígida**: ou a música passa, ou não passa. Uma música de energia 0,69 ficaria de fora de um pedido de 0,7, mesmo sendo quase igual. Por isso o `recomendar` usa **pontos**: dá uma **nota** a cada música e escolhe as de maior nota. Vamos entender como essa nota é calculada.

### A nota de cena: a música está perto do que a pessoa quer?

Imagine uma **régua de 0 a 1** para a energia. O pedido quer **0,85**: marcamos um **X** na régua. Cada música tem a sua marquinha. Medimos a **distância** entre o X e a marquinha de cada música:

| Música | Energia | Quer | Distância | Nota (1 − distância) |
|---|---|---|---|---|
| A | 0,90 | 0,85 | 0,05 (pertinho) | **0,95** |
| B | 0,60 | 0,85 | 0,25 | **0,75** |
| C | 0,10 | 0,85 | 0,75 (longe) | **0,25** |

**Perto = nota alta. Longe = nota baixa.** Fazemos a mesma conta com a **positividade** (o quão alegre a música soa) e tiramos a **média** das duas notas. A distância é aquele `abs` do Bloco 0: a diferença **sem sinal**.

### A nota de contexto: a música combina com o que a pessoa pediu?

A nota de cena olhou para **números** (energia e positividade). A nota de **contexto** olha para **etiquetas**: a ocasião, o humor e o gênero. E aqui a pergunta é de **sim ou não**.

Pense numa **lista de conferência**. Para cada coisa que a pessoa pediu, a música ganha:
- **1** se combina ✔
- **0** se não combina ✘

A nota é a **média** dos pontos. Exemplo: o pedido é **treino**, humor **empoderado**, gênero **pop**.

| Música | Ocasião (treino?) | Humor (empoderado?) | Gênero (pop?) | Conta | Nota |
|---|---|---|---|---|---|
| A | ✔ 1 | ✔ 1 | ✔ 1 | (1 + 1 + 1) / 3 | **1,00** |
| B | ✔ 1 | ✘ 0 | ✘ 0 | (1 + 0 + 0) / 3 | **0,33** |
| C | ✘ 0 | ✘ 0 | ✘ 0 | (0 + 0 + 0) / 3 | **0,00** |

Só entram na conta as coisas que a pessoa **pediu**. Se o pedido não tiver gênero, a lista tem só 2 itens e a média é dividida por 2.

A pergunta da **ocasião** é a mesma que você fez à mão: `.str.contains('treino')`. A do **humor** e a do **gênero** é "é **igual**?" (os dois iguais).

````python
PESOS = {'cena': 0.50, 'contexto': 0.35, 'qualidade': 0.15}   # quanto cada nota vale (somam 1)


def nota_cena(musicas, pedido):
    """Perto do que a pessoa quer = nota alta. Longe = nota baixa."""
    longe_energia = (musicas['energia'] - pedido.get('energia', 0.5)).abs()
    longe_positividade = (musicas['positividade'] - pedido.get('positividade', 0.5)).abs()
    return 1 - (longe_energia + longe_positividade) / 2


def nota_contexto(musicas, pedido):
    """Lista de conferência: 1 ponto para cada coisa pedida que a música cumpre."""
    pontos = []
    if 'ocasiao' in pedido:
        pontos.append(musicas['ocasioes_musica'].str.contains(pedido['ocasiao']).astype(float))
    if 'humor' in pedido:
        pontos.append((musicas['humor_predominante'] == pedido['humor']).astype(float))
    if 'genero' in pedido:
        pontos.append((musicas['genero_principal'] == pedido['genero']).astype(float))
    if not pontos:
        return pd.Series(0.5, index=musicas.index)   # o pedido não trouxe nada: nota neutra
    return sum(pontos) / len(pontos)
````

### Teste de mesa: conferindo as notas de 5 músicas

Vamos olhar a nota de **5 músicas** para um pedido de treino. Testar uma peça pequena antes de juntar tudo é o hábito que separa quem programa de quem sofre.

````python
amostra = musicas.head(5).copy()
pedido_teste = {'energia': 0.85, 'positividade': 0.65, 'ocasiao': 'treino', 'humor': 'empoderado'}

amostra['nota_cena'] = nota_cena(amostra, pedido_teste).round(2)
amostra['nota_contexto'] = nota_contexto(amostra, pedido_teste).round(2)
amostra[['titulo', 'energia', 'positividade', 'nota_cena', 'ocasioes_musica', 'humor_predominante', 'nota_contexto']]
````

### A função `recomendar`: tudo junto

Agora a função que monta a playlist a partir de um pedido. Ela faz **os mesmos passos do DJ**:

1. **Começar** com todas as músicas
2. **Peneirar**: país e gênero (só se o pedido trouxe)
3. **Dar a nota** a cada música (cena, contexto e qualidade)
4. **Escolher** as de maior nota (`sort_values` e `head`, que você já usou!)
5. **Ordenar** por intensidade: começa calma e vai esquentando, como um DJ de verdade

Você **não precisa escrever** esta função. Leia com calma e reconheça o que já sabe.

````python
def recomendar(pedido, quantidade=8):
    """Recebe um pedido (dicionário) e devolve a playlist."""
    # 1) começa com todas as músicas
    candidatas = musicas.copy()

    # 2) peneira: só se o pedido trouxe país ou gênero
    if 'pais' in pedido:
        candidatas = candidatas[candidatas['pais_artista'] == pedido['pais']]
    if 'genero' in pedido:
        do_genero = candidatas[candidatas['genero_principal'] == pedido['genero']]
        if len(do_genero) >= quantidade:      # só peneira se sobrar música suficiente
            candidatas = do_genero

    # 3) dá a nota a cada música: cena + contexto + qualidade, cada uma com o seu peso
    nota = (PESOS['cena'] * nota_cena(candidatas, pedido)
            + PESOS['contexto'] * nota_contexto(candidatas, pedido)
            + PESOS['qualidade'] * candidatas['score_base'])
    candidatas = candidatas.assign(nota=nota)

    # 4) escolhe as de maior nota
    playlist = candidatas.sort_values('nota', ascending=False).head(quantidade)

    # 5) ordena por intensidade: da mais calma para a mais forte
    playlist = playlist.sort_values('intensidade')
    return playlist[['titulo', 'artista', 'genero_principal', 'energia', 'intensidade', 'nota', 'link_spotify']].round(2)
````

## Bloco 6 — O agente: entender o texto da pessoa

Até aqui, o pedido era um dicionário que **nós** escrevíamos à mão. Mas ninguém vai escrever `{'energia': 0.85, 'ocasiao': 'treino'}` num app. A pessoa vai escrever uma **frase**: *"quero treinar de manhã com a galera"*.

Quem transforma a frase no dicionário é o **agente**.

### A analogia do garçom 🍽️

Num restaurante, o cliente diz *"quero algo leve, sem glúten, para dividir"*. O **garçom** entende e escreve na comanda o que a **cozinha** precisa: *"salada, sem glúten, 2 pratos"*. O garçom **não cozinha**. Ele só **traduz**.

| No restaurante | No nosso projeto |
|---|---|
| Cliente | a pessoa que escreve a frase |
| Garçom | o **Gemini**, a IA do Google (traduz a frase em pedido) |
| Comanda | o **pedido** (o dicionário) |
| Cozinha | o `recomendar`, que monta a playlist |

> **A regra de ouro do projeto:** a IA não escolhe músicas. Ela só traduz o texto em pedido. Quem consulta a base e monta a playlist é o nosso código.

Por quê? Uma IA que "sugere músicas" pode **inventar**: uma música que não existe, ou um artista errado. Assim, o resultado vem **sempre da nossa base**, de músicas reais, com link.

### Peça 1 — O cardápio e o segurança

O garçom só pode anotar o que **existe** no cardápio. Se o cliente pedir "uma ocasião chamada dormir", mas a nossa base não tem essa ocasião, o pedido não pode passar.

Por isso temos duas coisas:

1. **O cardápio**: as listas do que é válido (ocasiões, humores, países e gêneros).
2. **O segurança** (`validar_pedido`): confere o pedido que a IA devolveu, como o segurança de uma balada confere a lista na porta. **Corrige** o que dá (uma energia de 1,7 vira 1,0) e **barra** o que não serve (a ocasião "dormir").

> **Nunca confie na resposta de uma IA sem conferir.**

````python
OCASIOES = ['festa', 'estrada', 'treino', 'estudar', 'relaxar', 'sofrencia',
            'churrasco', 'romantico', 'limpar_casa', 'pre_balada']
HUMORES = ['animado', 'calmo', 'melancolico', 'romantico', 'nostalgico', 'empoderado']
PAISES = ['BR', 'INT']
GENEROS = sorted(musicas['genero_principal'].unique())    # os gêneros que existem na base

print('Gêneros disponíveis:', GENEROS)
````

````python
def validar_pedido(bruto):
    """O segurança: confere o pedido que veio (do Gemini ou de qualquer lugar) e devolve só o que é válido."""
    pedido = {}

    # energia e positividade: têm que ser números. Se passar de 1 ou ficar abaixo de 0, corrige.
    for campo in ['energia', 'positividade']:
        valor = bruto.get(campo)
        if isinstance(valor, (int, float)):                      # é um número?
            pedido[campo] = min(1.0, max(0.0, valor))            # prende entre 0 e 1

    # ocasião, humor, país e gênero: só passam se estiverem no cardápio
    cardapio = {'ocasiao': OCASIOES, 'humor': HUMORES, 'pais': PAISES, 'genero': GENEROS}
    for campo, validos in cardapio.items():
        if bruto.get(campo) in validos:
            pedido[campo] = bruto[campo]

    return pedido


# teste: valores absurdos viram valores seguros
print(validar_pedido({'energia': 1.7, 'positividade': 'alta', 'ocasiao': 'dormir', 'humor': 'animado', 'genero': 'sertanejo'}))
````

### O tradutor: a função que chama o Gemini

Agora a função que leva o prompt até o Gemini e traz a **comanda** (o pedido) de volta. Ela é uma **caixa-preta**: você só precisa saber que **entra uma frase e sai um pedido**.

Duas coisas que ela faz, em palavras simples:

- **Formulário (schema):** em vez de deixar o Gemini responder do jeito que quiser, a função entrega um **formulário** com campos e opções permitidas. O Gemini é obrigado a preencher o formulário.
- **Cinto e suspensório:** mesmo assim, a resposta passa pelo **segurança** (`validar_pedido`).

Rode a célula. Ela só ensina a função, não mostra nada.

````python
def interpretar_com_gemini(texto):
    """Envia a frase ao Gemini e devolve o pedido (já conferido pelo segurança)."""
    from google import genai
    from google.genai import types
    from pydantic import BaseModel
    from typing import Literal, Optional

    class PedidoGemini(BaseModel):                                   # o formulário que o Gemini tem que preencher
        energia: Optional[float] = None
        positividade: Optional[float] = None
        ocasiao: Optional[Literal[tuple(OCASIOES)]] = None           # só aceita opções do cardápio
        humor: Optional[Literal[tuple(HUMORES)]] = None
        genero: Optional[Literal[tuple(GENEROS)]] = None
        pais: Optional[Literal[tuple(PAISES)]] = None

    load_dotenv()
    cliente = genai.Client(api_key=os.environ['GEMINI_API_KEY'],
                           http_options=types.HttpOptions(timeout=30000))   # espera no máximo 30 segundos
    instrucoes = PROMPT_GEMINI.format(ocasioes=', '.join(OCASIOES), humores=', '.join(HUMORES),
                                      generos=', '.join(GENEROS), texto=texto) + '\n' + REGRA_EXTRA
    resposta = cliente.models.generate_content(
        model=MODELO_GEMINI,
        contents=instrucoes,
        config=types.GenerateContentConfig(response_mime_type='application/json',
                                           response_schema=PedidoGemini),
    )
    return validar_pedido(resposta.parsed.model_dump())


def traduzir(frase):
    """Mostra o Gemini traduzindo uma frase em pedido (sem derrubar o caderno se ele falhar)."""
    print('Frase :', frase)
    try:
        print('Pedido:', interpretar_com_gemini(frase))
    except Exception as erro:
        print('O Gemini não respondeu (' + type(erro).__name__ + '). Sem problema: o plano B, logo adiante, cobre isso.')
````

### Peça 4 — O plano B: o para-quedas 🪂

E se o Gemini **falhar**? Pode ser a internet que caiu, a cota do dia que acabou, a chave errada. Um app de verdade **não pode quebrar** por isso.

Por isso existe o **plano B**: uma versão **simples**, que **não usa IA**, só procura **palavras-chave** na frase. Se achar `trein`, monta um pedido de treino. Se achar `pagod`, acrescenta o gênero pagode.

O dicionário `CENAS` liga **o começo de uma palavra** a um pedido pronto. Usamos o *começo* para pegar variações: `'trein'` casa com *treino*, *treinar*, *treinando*.

Um dicionário! Você já sabe mexer nele (Bloco 0).

**Exercício 39**
Acrescente **duas cenas suas** ao dicionário `CENAS`. Cada uma precisa de uma palavra-chave (o começo da palavra), `energia`, `positividade`, `ocasiao` e `humor`. As ocasiões válidas estão em `OCASIOES` e os humores em `HUMORES`. Ideias: `'faxina'`, `'viagem'`, `'churrasc'`, `'dormir'`...

````python
CENAS = {   # começo da palavra -> pedido ('trein' casa com treino, treinar, treinando...)
    'trein':   {'energia': 0.85, 'positividade': 0.65, 'ocasiao': 'treino', 'humor': 'empoderado'},
    'estud':   {'energia': 0.35, 'positividade': 0.40, 'ocasiao': 'estudar', 'humor': 'calmo'},
    'sofrenc': {'energia': 0.45, 'positividade': 0.25, 'ocasiao': 'sofrencia', 'humor': 'melancolico'},
    # Exercício 39: acrescente duas cenas suas aqui (não esqueça a vírgula no fim de cada linha)
}

print('Palavras-chave conhecidas:', list(CENAS))
````

````python
import unicodedata

def normalizar(texto):
    """Tira acentos, põe em minúsculas e tira espaços das pontas (a mesma limpeza do Bloco 2, em forma de função)."""
    texto = unicodedata.normalize('NFKD', str(texto))
    texto = ''.join(c for c in texto if not unicodedata.combining(c))
    return texto.lower().strip()


GENEROS_PALAVRAS = {'forr': 'forro', 'sertanej': 'sertanejo', 'pagod': 'pagode', 'samba': 'samba',
                    'funk': 'funk', 'rock': 'rock', 'pop': 'pop', 'rap': 'rap', 'mpb': 'mpb'}


def interpretar_sem_ia(texto):
    """Plano B: entende a frase procurando palavras-chave. Não usa internet nem IA."""
    palavras = normalizar(texto).split()              # a frase, limpa e quebrada em palavras
    pedido = {}

    for palavra in palavras:                          # olha cada palavra da frase...
        for comeco, cena in CENAS.items():            # ...e compara com cada cena do dicionário
            if palavra.startswith(comeco):            # a palavra começa igual? ("treinar" começa com "trein")
                pedido = dict(cena)                   # copia o pedido pronto daquela cena

    for palavra in palavras:                          # agora procura o gênero
        for comeco, genero in GENEROS_PALAVRAS.items():
            if palavra == comeco or (len(comeco) >= 4 and palavra.startswith(comeco)):   # palavras curtas (pop, rap) têm que ser iguais
                pedido['genero'] = genero

    if any(p.startswith('brasileir') for p in palavras):   # e o país
        pedido['pais'] = 'BR'

    return validar_pedido(pedido)                     # o segurança confere antes de devolver


print(interpretar_sem_ia('quero treinar de manhã, só músicas brasileiras'))
print(interpretar_sem_ia('uma playlist de pagode para o churrasco'))
````

### Peça 5 — O agente completo

Agora a função que **junta tudo**: `gerar_playlist`. Ela:

1. **tenta o Gemini** (o garçom);
2. **se der qualquer erro**, usa o **plano B** (o para-quedas);
3. entrega o pedido ao **recomendador** (a cozinha);
4. devolve a playlist e **de onde veio o pedido** (`origem`: `gemini` ou `plano B`).

Isso se chama **tolerância a falhas**: o app continua funcionando mesmo quando uma peça falha. Como numa viagem de Uber: *tento o Uber; se não vier, vou a pé*.

Para você ver o plano B funcionando **sem desligar a internet**, a função tem um botão: `usar_gemini=False`.

````python
def gerar_playlist(texto, quantidade=8, usar_gemini=True):
    """Frase -> pedido (Gemini, ou plano B se falhar) -> playlist. Devolve também de onde veio o pedido."""
    try:
        if not usar_gemini:
            raise RuntimeError('Gemini desligado')                 # força o plano B (modo de teste)
        pedido, origem = interpretar_com_gemini(texto), 'gemini'
    except Exception as erro:                                      # qualquer erro do Gemini cai aqui
        pedido, origem = interpretar_sem_ia(texto), f'plano B ({type(erro).__name__})'
    if not pedido:
        raise ValueError('Não entendi o pedido. Tente algo como "treino de manhã" ou "viagem com as amigas".')
    return pedido, recomendar(pedido, quantidade=quantidade), origem


def mostrar(texto, quantidade=8, usar_gemini=True):
    """Roda o agente e mostra: de onde veio o pedido, o pedido entendido e a playlist."""
    try:
        pedido, playlist, origem = gerar_playlist(texto, quantidade, usar_gemini)
    except ValueError as erro:
        print('😕', erro)
        return
    print('Origem :', origem)
    print('Pedido :', pedido)
    return playlist[['titulo', 'artista', 'genero_principal', 'energia', 'nota']]
````

Então falta um **balcão de atendimento**: um lugar com endereço fixo, onde o site bate e pergunta *"monta uma playlist para esta frase aqui"*. Esse balcão se chama **servidor**, e o endereço dele vai ser `http://127.0.0.1:8000` — que quer dizer "aqui mesmo, neste computador, na porta 8000".

A conversa inteira tem quatro passos:

O passo 3 é todo seu, você já fez. Os outros três são as células abaixo, e elas **já vêm prontas**: são só tradução e encanamento. Leia com calma, rode e veja o seu agente ganhar um site.

### Peça 1 — O tradutor 🔁

A sua `gerar_playlist` devolve uma **tabela** do pandas. O site não entende tabela: ele espera uma **lista de dicionários**, com os nomes em inglês, assim:

```
{'title': 'Evidências', 'artist': 'Chitãozinho & Xororó', 'links': {'spotify': 'https://...'}}
```

É a mesma música, só com outra roupa. A função abaixo faz essa troca de roupa: percorre a tabela **uma linha por vez** (um `for`, igual ao do Bloco 0) e vai montando a lista do jeito que o site quer.

A playlist também precisa de um **nome**, que aparece em cima da lista no site. Para isso usamos um **dicionário** que liga cada ocasião a um nome bonito. Mais um dicionário — a essa altura você já é íntima deles.

````python
NOMES_DE_PLAYLIST = {        # ocasião -> nome que aparece lá no site
    'treino':    'Playlist de treino',
    'estudar':   'Para estudar',
    'sofrencia': 'Modo sofrência',
    'festa':     'Pista liberada',
    'estrada':   'Na estrada',
    'relaxar':   'Para relaxar',
    'churrasco': 'Churrasco de domingo',
    'romantico': 'Modo romântico',
}


def traduzir_para_o_site(playlist, pedido):
    """A tabela do pandas vira a lista de músicas que o site entende."""
    musicas_do_site = []
    for _, musica in playlist.iterrows():        # percorre a tabela, uma linha por vez
        musicas_do_site.append({
            'title':  musica['titulo'],
            'artist': musica['artista'].replace(';', ', '),   # a base separa artistas com ;
            'links':  {'spotify': musica['link_spotify']},
        })
    nome = NOMES_DE_PLAYLIST.get(pedido.get('ocasiao'), 'Sua playlist de hoje')
    return {'title': nome, 'tracks': musicas_do_site}


# Teste: duas músicas, já com a roupa do site
pedido_teste, playlist_teste, _ = gerar_playlist('quero treinar de manhã', quantidade=2)
traduzir_para_o_site(playlist_teste, pedido_teste)
````

### Peça 2 — O balcão 🛎️

Agora o balcão em si. São poucas linhas, mas cada uma tem um papel:

- `app = FastAPI()` cria o balcão, vazio.
- `@app.post('/api/playlist')` é a **plaquinha** pendurada nele: avisa que, quando chegar uma mensagem nesse endereço, é a função logo abaixo que atende. (Essa linha com `@` se chama *decorador*; por hoje, pense nela só como uma plaquinha.)
- `def atender(mensagem)` é quem faz o trabalho: pega a frase, chama **a sua** `gerar_playlist` e devolve o resultado já traduzido.

Repare no `try` / `except`: é o mesmo cuidado do plano B. Se a pessoa escrever algo que o agente não entende, em vez de o site quebrar, ela recebe um **recado educado** na tela.

````python
from fastapi import FastAPI, HTTPException

app = FastAPI()                                # o balcão, ainda vazio


@app.post('/api/playlist')                     # a plaquinha: é aqui que o site bate
def atender(mensagem: dict):
    """Recebe a frase da pessoa e devolve a playlist pronta para o site."""
    texto = mensagem['message']                # a frase que a pessoa digitou no site

    try:
        pedido, playlist, origem = gerar_playlist(texto)      # a SUA função, a mesma de sempre
    except ValueError:
        raise HTTPException(status_code=400,
                            detail='Não entendi o pedido. Tente algo como "treino de manhã".')

    print('Pedido do site:', texto, '| origem:', origem)      # aparece aqui embaixo, ao vivo

    return {
        'reply': f'Separei {len(playlist)} músicas para você.',
        'playlist': traduzir_para_o_site(playlist, pedido),
    }
````

### Peça 3 — Ligar o balcão 🔌

Falta abrir as portas. Um detalhe atrapalha: um servidor fica **esperando** mensagens para sempre, sem parar. Se a gente ligasse ele aqui direto, esta célula nunca terminaria e o caderno ficaria travado.

A solução é ligar o servidor numa **linha paralela** (o nome técnico é *thread*): ele fica atendendo de um lado, e o caderno continua livre do outro. É como pôr uma panela no fogo baixo e seguir cozinhando o resto do almoço.

````python
import threading
import time
import uvicorn


def ligar_o_balcao():
    """Liga o servidor numa linha paralela, para o caderno continuar livre."""
    configuracao = uvicorn.Config(app, host='127.0.0.1', port=8000, log_level='warning')
    servidor = uvicorn.Server(configuracao)
    threading.Thread(target=servidor.run, daemon=True).start()
    time.sleep(2)                              # dois segundos para ele acordar
    return servidor


servidor = ligar_o_balcao()

print('O seu agente está no ar em http://127.0.0.1:8000')
print()
print('Agora abra um terminal, entre na pasta frontend e rode:')
print('    npm run dev')
print()
print('Deixe este caderno aberto. Se reiniciar o kernel, o balcão fecha.')
````
